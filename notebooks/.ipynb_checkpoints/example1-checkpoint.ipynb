# Example 1

```python
import gepader
import utils
import glob
import spacy
from spacy import displacy
```

Specify input directory and read input file names

```python
indir = '../speeches'
jfiles = glob.glob(indir + '/*.json')
speeches = []
```

Read input files and convert them to Gepader objects

```python
for jfile in jfiles: 
    with open(jfile, 'r') as inf: 
        speeches.append(gepader.Gepader.from_json(inf.read())) 
```

Extract content for each speech

```python
content_dict = utils.extract_content(speeches)
```

Extract metadata for each speech

```python
meta_dict = utils.extract_metadata(speeches)
```

Search for specific words and return their token ids as a list

```python
word = 'Gleichstellung'
word_ids = utils.search_for(speeches, word)
layers = ['SPEECHACT', 'MORAL']
# Get all annotations (if any) for the annotation layers and the specified word
annotations = utils.get_annotations_by_id(word_ids, speeches, layers)
```

```python
annot_dict = gepader.Gepader.display_keyword_with_annots(annotations, meta_dict, word, layers, 'display')
```
